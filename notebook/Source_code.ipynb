```python
# IQR Calculation
def check_outlier_iqr(column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]

    print(f"--- {column} ---")
    print(f"Q1: {Q1}")
    print(f"Q3: {Q3}")
    print(f"IQR: {IQR}")
    print(f"Lower bound: {lower_bound}")
    print(f"Upper bound: {upper_bound}")
    print(f"Jumlah outlier: {len(outliers)}")
    print()
    return outliers

income_outliers = check_outlier_iqr('Income')
print(income_outliers[['ID', 'Income']])

print("="*50)

birth_outliers = check_outlier_iqr('Year_Birth')
print(birth_outliers[['ID', 'Year_Birth']])
```

#**ANOVA & Kruskal-Wallis**

```python
from scipy import stats

# Check normality of Conversion_Rate within each Age_Group using Shapiro-Wilk test
print('--- Shapiro-Wilk Normality Test per Age Group ---')
for group in df['Age_Group'].cat.categories:
    data = df[df['Age_Group'] == group]['Conversion_Rate']
    stat, p_value = stats.shapiro(data)
    print(f'{group}: p-value = {p_value:.6f}',
          '(NOT normal)' if p_value < 0.05 else '(normal)')
```

```python
# Prepare data: split Conversion_Rate by Age_Group
groups = [df[df['Age_Group'] == group]['Conversion_Rate'] for group in df['Age_Group'].cat.categories]

# ANOVA test
anova_stat, anova_p = stats.f_oneway(*groups)
print("--- ANOVA ---")
print(f"F-statistic: {anova_stat:.4f}")
print(f"p-value: {anova_p:.6f}")
print("Significant" if anova_p < 0.05 else "Not significant")

print()

# Kruskal-Wallis test
kw_stat, kw_p = stats.kruskal(*groups)
print("--- Kruskal-Wallis ---")
print(f"H-statistic: {kw_stat:.4f}")
print(f"p-value: {kw_p:.6f}")
print("Significant" if kw_p < 0.05 else "Not significant")
```

### Statistical Significance Testing — Age Group vs Conversion Rate

**Why ANOVA and Kruskal-Wallis?**

To test whether a numeric variable (Conversion_Rate) differs significantly
across multiple categorical groups (the five Age Groups), a test designed for
"category vs numeric" comparisons is needed, this rules out methods like linear
regression or R², which are meant for numeric-vs-numeric relationships (
Income vs Spending), not categorical groupings vs a continuous outcome.

ANOVA (Analysis of Variance) is the standard method for comparing the average of
a numeric variable across three or more groups. However, ANOVA assumes the data
within each group is normally distributed. Kruskal-Wallis is the non-parametric
counterpart to ANOVA, it tests the same type of question (do groups differ
significantly?) but does not require the normality assumption, making it more
robust when that assumption is violated.

Both tests were run in parallel so the results could be cross-checked against
each other, rather than relying on a single method.

To decide which result to trust, a Shapiro-Wilk normality test was run first on
Conversion_Rate within each Age Group.

**Normality check (Shapiro-Wilk test):**
All five groups returned p-values of 0.000000, indicating that Conversion_Rate
is not normally distributed in any group. This is expected given the
zero-inflated nature of the metric observed earlier in the distribution
analysis. Since the normality assumption behind ANOVA is violated, Kruskal-Wallis
is the more appropriate and trustworthy test for this dataset.

**Test results:**

| Test | Statistic | p-value | Result |
|---|---|---|---|
| ANOVA | F = 2.4083 | 0.047 | Significant (p < 0.05) |
| Kruskal-Wallis | H = 4.7240 | 0.317 | Not significant (p > 0.05) |

**Interpretation:**
The two tests disagree. Because the normality assumption behind ANOVA is
violated (confirmed by Shapiro-Wilk), the ANOVA result (p = 0.047) is unreliable,
even though it technically falls below the 0.05 threshold. Kruskal-Wallis result
(p = 0.317) is considered the more trustworthy conclusion here.

**Conclusion:** There is not enough statistical evidence to conclude that Age
Group has a significant relationship with Conversion Rate. The differences
observed earlier in average Conversion Rate across age groups (e.g. Youth at
0.062 vs Middle Age at 0.035) are likely due to random variation rather than a
true underlying pattern — particularly given the uneven sample sizes across
groups (Youth had only 116 customers, the smallest group by far).

```python
# Drop columns not needed for clustering:
# - ID: unique identifier, no behavioral signal
# - Z_CostContact, Z_Revenue: constant values across all rows, no variance
# - Dt_Customer, Year_Birth: already captured in the Age feature (redundant)
df_ml = df.drop(columns=['ID', 'Z_CostContact', 'Z_Revenue', 'Dt_Customer', 'Year_Birth'])

print("Shape after dropping columns:", df_ml.shape)
print(df_ml.columns.tolist())
```

```python
# Ordinal encoding: Education (SMA -> D3 -> S1 -> S2 -> S3)
education_order = {'SMA': 0, 'D3': 1, 'S1': 2, 'S2': 3, 'S3': 4}
df_ml['Education_Encoded'] = df_ml['Education'].map(education_order)

# Ordinal encoding: Age_Group (Youth -> Young Adult -> Adult -> Middle Age -> Senior)
age_group_order = {'Youth': 0, 'Young Adult': 1, 'Adult': 2, 'Middle Age': 3, 'Senior': 4}
df_ml['Age_Group_Encoded'] = df_ml['Age_Group'].map(age_group_order)

# Nominal encoding: Marital_Status (no natural order, use one-hot encoding)
df_ml = pd.get_dummies(df_ml, columns=['Marital_Status'], prefix='Marital')

# Drop original text columns now that encoded versions exist
df_ml = df_ml.drop(columns=['Education', 'Age_Group'])

print(df_ml.columns.tolist())
print("\nShape after encoding:", df_ml.shape)
df_ml.head()
```

```python
from sklearn.preprocessing import StandardScaler

# Columns that need standardization (continuous numeric, varying scales)
cols_to_scale = [
    'Income', 'Recency', 'MntCoke', 'MntFruits', 'MntMeatProducts',
    'MntFishProducts', 'MntSweetProducts', 'MntGoldProds',
    'NumDealsPurchases', 'NumWebPurchases', 'NumCatalogPurchases',
    'NumStorePurchases', 'NumWebVisitsMonth', 'Total_Spending',
    'Age', 'Total_Children', 'Total_Transactions', 'Conversion_Rate',
    'Education_Encoded', 'Age_Group_Encoded'
]

scaler = StandardScaler()

# Create a copy so the original unscaled values are preserved in df_ml
df_ml_scaled = df_ml.copy()
df_ml_scaled[cols_to_scale] = scaler.fit_transform(df_ml[cols_to_scale])

print(df_ml_scaled[cols_to_scale].describe().loc[['mean', 'std']])
df_ml_scaled.head()
```

#**Data Modeling**

```python
from sklearn.cluster import KMeans

# Compute WCSS (inertia) for k = 1 to 10
wcss = []
k_range = range(1, 11)

for k in k_range:
    kmeans = KMeans(n_clusters=k, init='k-means++', random_state=42, n_init=10)
    kmeans.fit(df_ml_scaled)
    wcss.append(kmeans.inertia_)

# Plot the Elbow curve
plt.figure(figsize=(8, 5))
plt.plot(k_range, wcss, marker='o')
plt.title('Elbow Method for Optimal k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('WCSS (Inertia)')
plt.xticks(k_range)
plt.show()
```

```python
from sklearn.metrics import silhouette_score

# Compute Silhouette Score for k = 2 to 10 (k=1 is not valid for this metric)
silhouette_scores = []
k_range_sil = range(2, 11)

for k in k_range_sil:
    kmeans = KMeans(n_clusters=k, init='k-means++', random_state=42, n_init=10)
    cluster_labels = kmeans.fit_predict(df_ml_scaled)
    score = silhouette_score(df_ml_scaled, cluster_labels)
    silhouette_scores.append(score)
    print(f"k={k}: Silhouette Score = {score:.4f}")

# Plot Silhouette Score per k
plt.figure(figsize=(8, 5))
plt.plot(k_range_sil, silhouette_scores, marker='o', color='green')
plt.title('Silhouette Score for Different k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Silhouette Score')
plt.xticks(k_range_sil)
plt.show()
```

```python
# Final KMeans clustering with k=3
kmeans_final = KMeans(n_clusters=3, init='k-means++', random_state=42, n_init=10)
cluster_labels = kmeans_final.fit_predict(df_ml_scaled)

# Attach cluster labels to both the scaled and original (unscaled) dataframes
df_ml_scaled['Cluster'] = cluster_labels
df_ml['Cluster'] = cluster_labels

# Quick check: how many customers fall into each cluster
print("--- Cluster distribution ---")
print(df_ml['Cluster'].value_counts().sort_index())

# Final Silhouette Score for k=3 (for reporting)
final_score = silhouette_score(df_ml_scaled.drop(columns=['Cluster']), cluster_labels)
print(f"\nFinal Silhouette Score (k=3): {final_score:.4f}")
```

```python
# Profile each cluster using key original (unscaled) columns
profile_cols = ['Income', 'Total_Spending', 'Total_Transactions', 'Age',
                 'Total_Children', 'Conversion_Rate', 'NumWebVisitsMonth', 'Response']

cluster_profile = df_ml.groupby('Cluster')[profile_cols].mean().round(2)
print(cluster_profile)
```

```python
from sklearn.decomposition import PCA

# Reduce to 2 dimensions for visualization purposes only
pca = PCA(n_components=2, random_state=42)
pca_result = pca.fit_transform(df_ml_scaled.drop(columns=['Cluster']))

df_ml_scaled['PCA1'] = pca_result[:, 0]
df_ml_scaled['PCA2'] = pca_result[:, 1]

print(f"Variance explained by 2 components: {sum(pca.explained_variance_ratio_):.2%}")

# Plot clusters in 2D PCA space
plt.figure(figsize=(9, 6))
colors = ['#E74C3C', '#3498DB', '#2ECC71']
for cluster_id in sorted(df_ml_scaled['Cluster'].unique()):
    subset = df_ml_scaled[df_ml_scaled['Cluster'] == cluster_id]
    plt.scatter(subset['PCA1'], subset['PCA2'], label=f'Cluster {cluster_id}',
                alpha=0.5, color=colors[cluster_id])

plt.title('Customer Clusters (PCA 2D Projection)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.show()
```

## Data Modeling — Customer Segmentation (K-Means Clustering)

### 1. Determining the Optimal Number of Clusters

**Elbow Method:** WCSS (inertia) was computed for k = 1 to 10. The curve shows a sharp
drop from k=1 to k=2, a moderate drop to k=3, and then flattens noticeably from k=3
onward — indicating diminishing returns beyond k=3.

**Silhouette Score:** Computed for k = 2 to 10 as a more objective cross-check.

| k | Silhouette Score |
|---|---|
| 2 | 0.2859 (highest) |
| 3 | 0.2071 |
| 4–10 | 0.11–0.13 (flat, low) |

k=2 technically has the highest score, but the gap to k=3 is small, and scores from
k=4 onward drop and stay flat — meaning k=2 and k=3 are the only reasonable candidates.

**Decision: k=3.** While k=2 has a marginally higher Silhouette Score, k=3 is supported
by the Elbow Method, remains close in Silhouette performance, and provides more
actionable granularity for marketing segmentation than a simple 2-way split.

**Caveat:** All Silhouette Scores obtained (max 0.29) fall below the 0.5 threshold
typically associated with strong, well-separated clusters. Scores in the 0.25–0.5
range indicate a weak but present cluster structure. This means the resulting
segments should be read as general tendencies rather than sharply distinct groups.

### 2. Final Model

K-Means was fitted with k=3 (`init='k-means++'`, `random_state=42`, `n_init=10`).

Final Silhouette Score: **0.2071**

### 3. Cluster Visualization (PCA)

A 2D PCA projection was used to visualize the clusters. The three groups appear

**Caveat:** the 2 components shown only explain **48.38%** of total variance in the
