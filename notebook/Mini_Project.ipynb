```python
# IQR Calculation
def check_outlier_iqr(column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]

    print(f"--- {column} ---")
    print(f"Q1: {Q1}")
    print(f"Q3: {Q3}")
    print(f"IQR: {IQR}")
    print(f"Lower bound: {lower_bound}")
    print(f"Upper bound: {upper_bound}")
    print(f"Jumlah outlier: {len(outliers)}")
    print()
    return outliers

income_outliers = check_outlier_iqr('Income')
print(income_outliers[['ID', 'Income']])

print("="*50)

birth_outliers = check_outlier_iqr('Year_Birth')
print(birth_outliers[['ID', 'Year_Birth']])
```

**1. Majority of customers did not convert (Conversion Rate = 0.0)**

these customers did not respond to the last marketing campaign, regardless of
how many times they visited the website. Since Response = 0 for these customers,
the formula (Response / NumWebVisitsMonth) always results in 0, no matter how
active they were on the website.

**2. A small group converted, but needed multiple visits (Conversion Rate between 0.1–0.5)**

The remaining customers who did respond to the campaign (Response = 1) are spread
across various Conversion Rate values, mostly between 0.1 and 0.5. This range
simply reflects how many visits it took before they finally responded — a
customer who visited 10 times before responding gets a lower rate (1/10 = 0.1),
while one who visited only 2 times before responding gets a higher rate (1/2 = 0.5).
The exact value doesn't indicate "how convinced" they were, only how many visits
were needed before conversion happened.

**3. A small number of customers converted immediately (Conversion Rate = 1.0)**

30 customers converted on their very first visit (NumWebVisitsMonth = 1,
Response = 1), representing the most "efficient" conversion possible under this
metric. This may suggest these customers were already influenced by factors
outside the website (word-of-mouth, social media, other channels) before visiting,
though the dataset cannot confirm this directly — noted as a limitation of the metric.

**Overall pattern:** The distribution is heavily right-skewed, since campaign
response itself is a rare event in this dataset (only ~15% of all customers
responded to the last campaign, based on the raw Response column).

#**ANOVA & Kruskal-Wallis**

```python
from scipy import stats

# Check normality of Conversion_Rate within each Age_Group using Shapiro-Wilk test
print('--- Shapiro-Wilk Normality Test per Age Group ---')
for group in df['Age_Group'].cat.categories:
    data = df[df['Age_Group'] == group]['Conversion_Rate']
    stat, p_value = stats.shapiro(data)
    print(f'{group}: p-value = {p_value:.6f}',
          '(NOT normal)' if p_value < 0.05 else '(normal)')
```

```python
# Prepare data: split Conversion_Rate by Age_Group
groups = [df[df['Age_Group'] == group]['Conversion_Rate'] for group in df['Age_Group'].cat.categories]

# ANOVA test
anova_stat, anova_p = stats.f_oneway(*groups)
print("--- ANOVA ---")
print(f"F-statistic: {anova_stat:.4f}")
print(f"p-value: {anova_p:.6f}")
print("Significant" if anova_p < 0.05 else "Not significant")

print()

# Kruskal-Wallis test
kw_stat, kw_p = stats.kruskal(*groups)
print("--- Kruskal-Wallis ---")
print(f"H-statistic: {kw_stat:.4f}")
print(f"p-value: {kw_p:.6f}")
print("Significant" if kw_p < 0.05 else "Not significant")
```

### Statistical Significance Testing — Age Group vs Conversion Rate

**Why ANOVA and Kruskal-Wallis?**

To test whether a numeric variable (Conversion_Rate) differs significantly
across multiple categorical groups (the five Age Groups), a test designed for
"category vs numeric" comparisons is needed, this rules out methods like linear
regression or R², which are meant for numeric-vs-numeric relationships (
Income vs Spending), not categorical groupings vs a continuous outcome.

ANOVA (Analysis of Variance) is the standard method for comparing the average of
a numeric variable across three or more groups. However, ANOVA assumes the data
within each group is normally distributed. Kruskal-Wallis is the non-parametric
counterpart to ANOVA, it tests the same type of question (do groups differ
significantly?) but does not require the normality assumption, making it more
robust when that assumption is violated.

Both tests were run in parallel so the results could be cross-checked against
each other, rather than relying on a single method.

To decide which result to trust, a Shapiro-Wilk normality test was run first on
Conversion_Rate within each Age Group.

**Normality check (Shapiro-Wilk test):**
All five groups returned p-values of 0.000000, indicating that Conversion_Rate
is not normally distributed in any group. This is expected given the
zero-inflated nature of the metric observed earlier in the distribution
analysis. Since the normality assumption behind ANOVA is violated, Kruskal-Wallis
is the more appropriate and trustworthy test for this dataset.

**Test results:**

| Test | Statistic | p-value | Result |
|---|---|---|---|
| ANOVA | F = 2.4083 | 0.047 | Significant (p < 0.05) |
| Kruskal-Wallis | H = 4.7240 | 0.317 | Not significant (p > 0.05) |

**Interpretation:**
The two tests disagree. Because the normality assumption behind ANOVA is
violated (confirmed by Shapiro-Wilk), the ANOVA result (p = 0.047) is unreliable,
even though it technically falls below the 0.05 threshold. Kruskal-Wallis result
(p = 0.317) is considered the more trustworthy conclusion here.

**Conclusion:** There is not enough statistical evidence to conclude that Age
Group has a significant relationship with Conversion Rate. The differences
observed earlier in average Conversion Rate across age groups (e.g. Youth at
0.062 vs Middle Age at 0.035) are likely due to random variation rather than a
true underlying pattern — particularly given the uneven sample sizes across
groups (Youth had only 116 customers, the smallest group by far).

#**Total Children x Conversion Rate**

```python
# Average Conversion Rate per Total_Children group
avg_cr_by_children = df.groupby('Total_Children')['Conversion_Rate'].mean()
print("--- Average Conversion Rate by Total_Children ---")
print(avg_cr_by_children)

sample_size_children = df['Total_Children'].value_counts().sort_index()
print("\n--- Sample size per Total_Children group ---")
print(sample_size_children)

print("\n" + "="*50)

# Kruskal-Wallis test across Total_Children groups
groups_children = [df[df['Total_Children'] == n]['Conversion_Rate'] for n in sorted(df['Total_Children'].unique())]

kw_stat_children, kw_p_children = stats.kruskal(*groups_children)
print("--- Kruskal-Wallis: Total_Children vs Conversion_Rate ---")
print(f"H-statistic: {kw_stat_children:.4f}")
print(f"p-value: {kw_p_children:.6f}")
print("Significant" if kw_p_children < 0.05 else "Not significant")
```

```python
# Bar chart: Average Conversion Rate by Total_Children (unrounded labels)
avg_cr_by_children = df.groupby('Total_Children')['Conversion_Rate'].mean()

plt.figure(figsize=(8, 5))
bars = plt.bar(avg_cr_by_children.index.astype(str), avg_cr_by_children.values, edgecolor='black')

# Add value labels on top of each bar (6 decimal places, matching raw data)
for bar, value in zip(bars, avg_cr_by_children.values):
    plt.text(bar.get_x() + bar.get_width()/2, value, f'{value:.6f}',
              ha='center', va='bottom', fontsize=9)

plt.title('Average Conversion Rate by Total_Children')
plt.xlabel('Total Children')
plt.ylabel('Average Conversion Rate')
plt.show()
```

|---|---|---|
| 0 | 0.115 | 635 |
| 1 | 0.015 | 1,124 |

**Interpretation:** Unlike Age Group, the relationship between Total_Children and

#**Total Spending x Conversion Rate**

```python
# Spearman correlation: Total_Spending vs Conversion_Rate
spearman_corr, spearman_p = stats.spearmanr(df['Total_Spending'], df['Conversion_Rate'])

print("--- Spearman Correlation: Total_Spending vs Conversion_Rate ---")
print(f"Correlation coefficient: {spearman_corr:.4f}")
print(f"p-value: {spearman_p:.6f}")
print("Significant" if spearman_p < 0.05 else "Not significant")
```

```python
# Scatter plot: Total_Spending vs Conversion_Rate
plt.figure(figsize=(8, 5))
plt.scatter(df['Total_Spending'], df['Conversion_Rate'], alpha=0.4)
plt.title('Total_Spending vs Conversion_Rate')
plt.xlabel('Total Spending')
plt.ylabel('Conversion Rate')
plt.show()
```

### Statistical Significance Testing — Total_Spending vs Conversion Rate

**Spearman correlation test:** rho = 0.2598, p-value ≈ 0.000000 → Significant

**Interpretation:** There is a statistically significant positive relationship
between Total_Spending and Conversion_Rate — customers who spend more overall
tend to have somewhat higher conversion rates. However, the strength of this
relationship is weak (rho = 0.26), meaning spending alone explains only a small
part of the variation in conversion behavior. This is consistent with the
correlation heatmap observed earlier (0.35).

The scatter plot shows conversion rate values forming distinct horizontal bands
(0.0, 0.1, 0.2, 0.33, 0.5, 1.0) rather than a continuous spread. This is expected,
since Conversion_Rate is calculated as Response divided by NumWebVisitsMonth, and
NumWebVisitsMonth only takes whole number values — so only a limited set of
possible rates can occur. Within these bands, customers in the higher rate bands
(0.5, 1.0) tend to skew slightly toward higher spending, which is consistent with
the weak positive correlation found in the test.
